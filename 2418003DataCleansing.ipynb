1. Import Libraries

```python
import pandas as pd
import numpy as np
import re
from IPython.display import display
```

2. Menghubungkan Google Drive

```python
from google.colab import drive

drive.mount('/content/drive')
```

3. Memuat Data Penjualan yang Belum Bersih

```python
file_path = '/content/drive/MyDrive/Colab Notebooks/Data kotor penjualan baju distro.xlsx'

data = pd.read_excel(
    file_path,
    sheet_name='Data Kotor',
    engine='openpyxl'
)

print("DATA PENJUALAN BAJU DISTRO")
display(data)
```

4. Melihat Informasi Dataset

```python
print("\nInformasi Dataset:")
data.info()

print("\nJenis Data Setiap Kolom:")
print(data.dtypes)
```

5. Merapikan Nama Produk

```python
def rapikan_produk(product):
    if pd.isna(product):
        return np.nan

    product = str(product).strip()
    product = re.sub(r'\s+', ' ', product)
    product = product.title()

    return product


data['Produk Bersih'] = data['Nama Produk'].apply(rapikan_produk)

print("\nPerbandingan Nama Produk:")
display(data[['Nama Produk', 'Produk Bersih']].head(15))
```

6. Menyeragamkan Penulisan Warna

```python
def rapikan_warna(color):
    if pd.isna(color):
        return np.nan

    color = str(color).strip().lower()
    color = re.sub(r'\s+', ' ', color)

    if color in ['abu abu', 'abu-abu']:
        return 'Abu-abu'

    return color.title()


data['Warna Bersih'] = data['Warna'].apply(rapikan_warna)

print("\nPerbandingan Data Warna:")
display(data[['Warna', 'Warna Bersih']].head(15))
```

7. Memperbaiki Format Ukuran Pakaian

```python
def rapikan_ukuran(size):
    if pd.isna(size):
        return np.nan

    size = str(size).strip().upper()

    return size


data['Ukuran Bersih'] = data['Ukuran'].apply(rapikan_ukuran)

print("\nPerbandingan Ukuran:")
display(data[['Ukuran', 'Ukuran Bersih']].head(15))
```

8. Validasi Nilai Harga

```python
def rapikan_harga(price):
    if pd.isna(price):
        return np.nan

    price = str(price).strip()
    price = price.replace('.', '').replace(',', '')

    try:
        price = float(price)

        if price <= 0:
            return np.nan

        return price

    except:
        return np.nan


data['Harga Bersih'] = data['Harga (Rp)'].apply(rapikan_harga)

print("\nPemeriksaan Harga:")
display(data[['Harga (Rp)', 'Harga Bersih']].head(15))
```

9. Mengoreksi Jumlah Produk Terjual

```python
def rapikan_jumlah(quantity):
    if pd.isna(quantity):
        return np.nan

    quantity = str(quantity).strip().lower()

    angka = {
        'satu': 1,
        'dua': 2,
        'tiga': 3,
        'empat': 4,
        'lima': 5,
        'enam': 6,
        'tujuh': 7,
        'delapan': 8,
        'sembilan': 9,
        'sepuluh': 10
    }

    if quantity in angka:
        return angka[quantity]

    try:
        quantity = int(float(quantity))

        if quantity <= 0:
            return np.nan

        return quantity

    except:
        return np.nan


data['Jumlah Terjual Bersih'] = (
    data['Jumlah Terjual'].apply(rapikan_jumlah)
)

print("\nPerbandingan Jumlah Terjual:")
display(data[['Jumlah Terjual', 'Jumlah Terjual Bersih']].head(15))
```

10. Menyesuaikan Format Tanggal

```python
data['Tanggal Bersih'] = pd.to_datetime(
    data['Tanggal'],
    errors='coerce',
    dayfirst=True
)

print("\nPerbandingan Tanggal:")
display(data[['Tanggal', 'Tanggal Bersih']].head(15))
```

11. Menata Status Pembayaran

```python
def rapikan_pembayaran(status):
    if pd.isna(status):
        return np.nan

    status = str(status).strip().lower()

    if status == 'lunas':
        return 'Lunas'

    elif status in ['belum', 'belum lunas']:
        return 'Belum Lunas'

    return np.nan


data['Pembayaran Bersih'] = (
    data['Status Pembayaran'].apply(rapikan_pembayaran)
)

print("\nPerbandingan Status Pembayaran:")
display(
    data[
        ['Status Pembayaran', 'Pembayaran Bersih']
    ].head(15)
)
```

12. Menyamakan Metode Penjualan

```python
data['Metode Bersih'] = (
    data['Metode Penjualan']
    .astype('string')
    .str.strip()
    .str.lower()
    .str.title()
)

print("\nPerbandingan Metode Penjualan:")
display(
    data[
        ['Metode Penjualan', 'Metode Bersih']
    ].head(15)
)
```

13. Menemukan Data yang Sama

```python
print("\nData yang Diduga Mengalami Pengulangan:")

duplicate_rows = data[
    data.duplicated(
        subset=[
            'Produk Bersih',
            'Warna Bersih',
            'Ukuran Bersih',
            'Tanggal Bersih'
        ],
        keep=False
    )
]

display(duplicate_rows)
```

14. Mengurangi Data Berulang

```python
data_deduplicated = data.drop_duplicates(
    subset=[
        'Produk Bersih',
        'Warna Bersih',
        'Ukuran Bersih',
        'Tanggal Bersih'
    ],
    keep='first'
)

data_deduplicated.reset_index(drop=True, inplace=True)

print("\nTotal Data Awal :", len(data))
print("Total Data Sesudah Pemeriksaan Duplikat :", len(data_deduplicated))
```

15. Memeriksa Nilai yang Hilang

```python
print("\nPengecekan Data Kosong:")
display(data_deduplicated.isnull().sum().to_frame('Jumlah Data Kosong'))
```

16. Mengatasi Harga yang Tidak Terisi

```python
median_price = data_deduplicated['Harga Bersih'].median()

data_deduplicated['Harga Bersih'] = (
    data_deduplicated['Harga Bersih']
    .fillna(median_price)
)

print("\nJumlah Data Harga yang Masih Kosong:")
print(data_deduplicated['Harga Bersih'].isnull().sum())
```

17. Melengkapi Jumlah Penjualan

```python
median_quantity = data_deduplicated['Jumlah Terjual Bersih'].median()

data_deduplicated['Jumlah Terjual Bersih'] = (
    data_deduplicated['Jumlah Terjual Bersih']
    .fillna(median_quantity)
)

print("\nJumlah Data Penjualan yang Masih Kosong:")
print(
    data_deduplicated['Jumlah Terjual Bersih'].isnull().sum()
)
```

18. Melengkapi Ukuran dan Tanggal

```python
if data_deduplicated['Ukuran Bersih'].isnull().any():
    mode_size = data_deduplicated['Ukuran Bersih'].mode()[0]

    data_deduplicated['Ukuran Bersih'] = (
        data_deduplicated['Ukuran Bersih']
        .fillna(mode_size)
    )


if data_deduplicated['Tanggal Bersih'].isnull().any():
    mode_date = data_deduplicated['Tanggal Bersih'].mode()[0]

    data_deduplicated['Tanggal Bersih'] = (
        data_deduplicated['Tanggal Bersih']
        .fillna(mode_date)
    )

print("\nSisa Data Kosong pada Ukuran:")
print(data_deduplicated['Ukuran Bersih'].isnull().sum())

print("\nSisa Data Kosong pada Tanggal:")
print(data_deduplicated['Tanggal Bersih'].isnull().sum())
```

19. Melengkapi Data Kategori

```python
if data_deduplicated['Warna Bersih'].isnull().any():
    mode_color = data_deduplicated['Warna Bersih'].mode()[0]

    data_deduplicated['Warna Bersih'] = (
        data_deduplicated['Warna Bersih']
        .fillna(mode_color)
    )


if data_deduplicated['Pembayaran Bersih'].isnull().any():
    mode_payment = data_deduplicated[
        'Pembayaran Bersih'
    ].mode()[0]

    data_deduplicated['Pembayaran Bersih'] = (
        data_deduplicated['Pembayaran Bersih']
        .fillna(mode_payment)
    )


if data_deduplicated['Metode Bersih'].isnull().any():
    mode_method = data_deduplicated[
        'Metode Bersih'
    ].mode()[0]

    data_deduplicated['Metode Bersih'] = (
        data_deduplicated['Metode Bersih']
        .fillna(mode_method)
    )


print("\nPemeriksaan Missing Value Terakhir:")
display(data_deduplicated.isnull().sum().to_frame('Jumlah Data Kosong'))
```

20. Menampilkan Dataset yang Sudah Dibersihkan

```python
hasil_cleansing = data_deduplicated[
    [
        'No',
        'Produk Bersih',
        'Warna Bersih',
        'Ukuran Bersih',
        'Harga Bersih',
        'Jumlah Terjual Bersih',
        'Tanggal Bersih',
        'Pembayaran Bersih',
        'Metode Bersih'
    ]
]

print("\n====================================================")
print("DATA PENJUALAN BAJU DISTRO SETELAH DIBERSIHKAN")
print("====================================================")

display(hasil_cleansing)
```

21. Cek Missing Value Setelah Cleansing

```python
print("\nMissing Value Setelah Data Cleansing:")
print(hasil_cleansing.isnull().sum())
```

22. Jumlah Data Terakhir

```python
print("\nJumlah Data Awal :", len(data))
print("Jumlah Data Akhir:", len(hasil_cleansing))
```

21. Mengekspor Hasil ke Excel

```python
output_path = '/content/drive/MyDrive/Colab Notebooks/Data Bersih Penjualan Baju Distro.xlsx'

hasil_cleansing.to_excel(
    output_path,
    index=False,
    sheet_name='Data Bersih'
)

print("\nFile data bersih berhasil dibuat.")
print("Lokasi file:")
print(output_path)
```
